```python
def clean_text(text):
    text = text.strip()
    text = text.lower()
    text = text.replace("-", " ")
    return text

text = "  Data-Science  "

print(clean_text(text))
```

```python
names = ["  harika ", "RAHUL", " priya  ", "  ANIL"]

clean_names = []

for name in names:
    name = name.strip()
    name = name.lower()
    name = name.title()
    clean_names.append(name)

print(clean_names)
```

```python
emails = [
    " harika@gmail.com ",
    "RAHUL@GMAIL.COM",
    " priya@gmail.com "
]

clean_emails = []

for email in emails:
    email = email.strip()
    email = email.lower()
    clean_emails.append(email)

print(clean_emails)
```

```python
categories = [" Male", "female ", "MALE", " Female", "male"]

clean_categories = []

for category in categories:
    category = category.strip().lower()
    clean_categories.append(category)

print(clean_categories)
```

```python
texts = ["Hello", " Hello", "World ", " Data Science "]

for text in texts:
    if text != text.strip():
        print("Unwanted spaces found:", repr(text))
    else:
        print("No unwanted spaces:", repr(text))
```

```python
customers = [
    {"name": "  Harika ", "email": " HARIKA@GMAIL.COM ", "category": " student "},
    {"name": "Rahul  ", "email": " rahul@gmail.com", "category": "STUDENT"},
    {"name": "  Priya", "email": "PRIYA@GMAIL.COM ", "category": " Student "}
]

for customer in customers:
    customer["name"] = customer["name"].strip().title()
    customer["email"] = customer["email"].strip().lower()
    customer["category"] = customer["category"].strip().lower()

print(customers)
```

```python
for customer in customers:
    print(customer)
```
